```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
llm = ChatGroq(
    model = "llama-3.1-8b-instant",
    temperature = 0.2,
    streaming = True
)
```

```python
@tool
def add_var(a:int , b:int):
    """
        It will return the sum of two Variable.
        
        Args:
        a: Variable 1
        b: Variable 2
    """
    return a+b

@tool
def multiply(a:int , b:int):
    """
    It will return the product of two variables.
    
    Args:
    a: variable 1
    b: variable 2
    """
    return a*b
```

```python
add_var.invoke({"a" : 12 , "b" : 8})
```

```python
agent = create_agent(
    model = llm,
    tools = [add_var ,multiply],
    system_prompt="You are a Math Teacher always use tools for Calculation"
)
```

```python
response = agent.invoke({
    "messages":[{"role" : "user" , "content" : "What is 2+3*20 ?."}]
    })

print(response['messages'][-1].content)
```
