```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from pydantic import BaseModel , Field
from typing import List
import json
```

```python
load_dotenv()
```

```python
llm = ChatGroq(
    model = "llama-3.3-70b-versatile",
    temperature = 0.2
)
```

```python
text = "Hello , My name is Hitesh Moota, Email is hiteshmoota1@gmail.com and my age is 27"
```

```python
class RequiredStruc(BaseModel):
    name:str = Field(description = "Name of the user")
    email:str = Field(description = "Email Address")
    age:int = Field(description = "Age")

structured_llm = llm.with_structured_output(RequiredStruc)
```

```python
res = structured_llm.invoke(text)
print(res.model_dump())
```

```python
class Movies(BaseModel):
    title:str = Field(description = "Movie Title")
    year:int = Field(description = "Movie Released Year.")

class AllMovies(BaseModel):
    movies:List[Movies]

movie_llm = llm.with_structured_output(AllMovies)
res = movie_llm.invoke("Give me 3 Sci-Fi Movie")
print(res.model_dump_json(indent = 2))
```
