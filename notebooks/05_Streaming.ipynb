```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import time
```

```python
load_dotenv()
```

```python
llm = ChatGroq(
    model = "llama-3.3-70b-versatile",
    temperature = 0.2,
    streaming = True
)
```

```python
text = "For AI Tech lead how many yrs it will take to reach to take if i start at the age of 28 yrs.Im already late i have never done any work unemployed"
```

```python
res = llm.stream(text)
for chunk in res:
    ans = chunk.content
    for char in ans:
        print(char , end = "" , flush = True)
        time.sleep(0.02)
```
