```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
```

```python
load_dotenv()
```

```python
llm = ChatGroq(
    model = "llama-3.3-70b-versatile",
    temperature=0.2
)
```

```python
history = []
```

```python
while True:
    input_query = input("You : ")

    if input_query.lower() == 'exit':
        print("Thank you for using ,Have a Nice Day")
        break
    
    print(f"you: {input_query}")
    history.append({"role" : "user" , "content" : input_query})
    res = llm.invoke(history)
    print(f"AI: {res.content}")
    history.append({"role" : "ai" , "content" : res.content})
    
```
