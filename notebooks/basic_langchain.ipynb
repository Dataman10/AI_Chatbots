```python
from langchain_groq import ChatGroq
prompt = [
    ("system" , "You are a Senior AI Engineer"),
    ("user" , "What Industries actually needs in an GenAI Developer or AI Engineer.Give me in bulleted points")
]

llm = ChatGroq(
    model = "llama-3.1-8b-instant",
    temperature= 0.2
)
res = llm.invoke(prompt)
print(res.content)
```
