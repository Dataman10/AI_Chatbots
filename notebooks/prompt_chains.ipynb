```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv


load_dotenv()
out = StrOutputParser()
```

```python
llm = ChatGroq(
    model = "llama-3.1-8b-instant",
    temperature = 0.2
)
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role" : "system" , "content" : "You are a Senior {status} Engineer"},
    {"role" : "user" , "content" : "{query}"}
])
```

```python
llm = ChatGroq(
    model = "llama-3.3-70b-versatile",
    temperature = 0.2
)
```

```python
chains = prompts | llm | out
chains
```

```python
res = chains.invoke({"status" : "AI" , "query" : "I have a gap year of 2 yrs and 27 yrs old ,if i join as an AI Engineer then How long it will take to be an Tech Lead"})
print(res)
```
